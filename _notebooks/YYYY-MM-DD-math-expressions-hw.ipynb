### Popcorn Hack 1: Try the Basics


```python
value_a = 100
value_b = 10
value_c = 25

addition = value_a + value_b + value_c
subtraction = value_a - value_b
multiplication = value_b * value_c
division = value_a / value_b

print("Total:", addition)
print("Difference:", subtraction)
print("Product:", multiplication)
print("Division:", division)
```

### Popcorn Hack 2: More Expressions


```python
base = 8
height = 5
side = 6

area_triangle = 0.5 * base * height
perimeter_square = 4 * side
average = (base + height + side) / 3

print("Triangle area:", area_triangle)
print("Square perimeter:", perimeter_square)
print("Average:", average)
```
