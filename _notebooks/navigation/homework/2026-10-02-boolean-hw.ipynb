### Popcorn Hack 3: SFI Search Filter

```python
# CODE_RUNNER: Popcorn Hack 3 - SFI Search Filter
query = "1.2"
records = [
    {"product_name": "Replacement Flywheels and Clutch Assemblies", "category": "Auto Racing", "spec_number": "1.1"},
    {"product_name": "Multiple Disc Clutch Assemblies", "category": "Drag Racing", "spec_number": "1.2"},
    {"product_name": "Racing Flywheel Record", "category": "Auto Racing", "spec_number": "2.1"}
]
valid_categories = ["Auto Racing", "Drag Racing"]

matches = []
for record in records:
    name_or_spec_match = query in record["product_name"] or query == record["spec_number"]
    category_ok = record["category"] in valid_categories
    if name_or_spec_match and category_ok:
        matches.append(record["product_name"])

print("Matches for query '1.2':", matches)  # Prediction: ['Multiple Disc Clutch Assemblies']
```

## MCQ

MCQ 3.05: 4/4 | answers: A,C,A,B

**Explanations:**
1. **1 - A (True):** Greater-than-or-equal includes equality, so `7 >= 7` is `True`.
2. **2 - C (and):** Both requirements (a name and a spec number) must be true at the same time, so `and` is the matching operator.
3. **3 - A (True):** `not` reverses a Boolean value, so `not False` is `True`.
4. **4 - B (`category == 'Auto Racing' or category == 'Drag Racing'`):** Accepting either category only requires one of the two comparisons to be true, which is what `or` does.

Test evidence: the validator accepted the one record that had every required field, an accepted category, and a brand-new spec number, and rejected the other two for two different reasons — an invalid category ("Street Car") and a duplicate spec number ("1.1") — which confirms the `and`/`not` logic matches the written rule for both a passing case and two different kinds of failing cases.
