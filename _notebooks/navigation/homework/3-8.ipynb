## Popcorn

```python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
# Theme: The SFSRC team scans wildfire alert levels from different zones.
# Goal: skip safe zones, count response zones, total their alert scores, and stop at the first critical zone.
alerts = [1, 4, 2, 6, 3, 9, 5]

# Alerts below this number are safe and should be skipped.
safe_cutoff = 3
# Alerts at or above this number are critical and should stop the scan.
critical_cutoff = 8

# Counters and totals start at 0 because no alerts have been processed yet.
action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    # If the alert is safe, count it, print a message, and skip to the next alert.
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    # If code reaches here, the alert needs action.
    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    # If this alert is critical, mark it and stop the scan.
    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

# Final report: these lines explain what happened during the scan.
print("Safe alerts skipped: " + str(safe_count))     # Prediction: 2 (alerts 1 and 2)
print("Action alerts: " + str(action_count))         # Prediction: 4 (alerts 4, 6, 3, 9)
print("Total action score: " + str(total_action_score))  # Prediction: 22 (4+6+3+9)
print("Critical found: " + str(critical_found))      # Prediction: True (9 >= 8 stops the scan before the last alert, 5, is ever reached)
```

## MCQ

MCQ 3.08 Iterations: 4/4 | answers: A,B,A,B

**Explanations:**
1. **1 - A (for loop):** A `for` loop is best when the sequence is already known, such as a list of readings.
2. **2 - B (Skips the rest of the current loop pass and moves to the next item):** `continue` skips the remaining statements for the current item only, then moves on to the next item.
3. **3 - A (A condition that eventually changes):** A `while` loop needs code inside the loop that can make its condition become false, or it never stops.
4. **4 - B (break):** `break` exits the loop immediately once the critical alert is found, which is exactly what stopping the scan early requires.

```python
# CODE_RUNNER: Homework Hack - themed loop-based checker
# Theme: Server load monitoring across a cluster of machines.
# Goal: skip safe load readings, count readings that need action, total their scores, and stop at the first critical reading.
server_loads = [12, 45, 30, 68, 15, 72, 50, 90, 40]

# Values below safe_cutoff should be skipped with continue.
safe_cutoff = 20

# Values at or above critical_cutoff should stop the loop with break.
critical_cutoff = 85

# Summary variables, all starting at 0/False since nothing has been checked yet.
checked_count = 0
action_count = 0
skipped_count = 0
total_score = 0
critical_found = False

# Traverse every value in the list until a critical value stops the scan.
for load in server_loads:
    # Count every value the loop looks at.
    checked_count += 1

    # Skip safe values, but count them before continuing.
    if load < safe_cutoff:
        skipped_count += 1
        print(str(load) + "%: safe, no action")
        continue

    # If code reaches here, this value needs action: count it and add it to the running total.
    action_count += 1
    total_score += load
    print(str(load) + "%: action needed")

    # Stop early if this value is critical.
    if load >= critical_cutoff:
        critical_found = True
        print("Critical load found. Stop checking.")
        break

# Final report.
print("Final Report")
print("Checked: " + str(checked_count))      # Prediction: 8 (the loop stops at the 8th value, 90; the 9th, 40, is never reached)
print("Skipped: " + str(skipped_count))      # Prediction: 2 (12% and 15%)
print("Actions: " + str(action_count))       # Prediction: 6 (45, 30, 68, 72, 50, 90)
print("Total score: " + str(total_score))    # Prediction: 355
print("Critical found: " + str(critical_found))  # Prediction: True (90 >= 85 stops the scan)
```
