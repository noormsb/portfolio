## Homework Hack
Refactored the signup form. No inline styles or made-up classes; every input uses `ocs__input` with a size modifier, and each has a visible label.

```python
%%html
<!-- UI_RUNNER: Inputs Homework Base -->

<label for="assistance-category">Assistance Category</label>
<input id="assistance-category" type="text" class="ocs__input large gradient" placeholder="Assistance Category">

<label for="volunteer-first-name">Volunteer First Name</label>
<input id="volunteer-first-name" type="text" class="ocs__input small" placeholder="Volunteer First Name">

<label for="volunteer-email">Volunteer Email</label>
<input id="volunteer-email" type="email" class="ocs__input medium" placeholder="Volunteer Email">
```

## What changed and why

| Original | Refactored | Why |
|---|---|---|
| `style="width: 500px; background: linear-gradient(...)"` | `ocs__input large gradient` | Size and gradient come from shared modifiers, not inline CSS |
| `class="tinybox"` | `ocs__input small` | `tinybox` was a made-up class; `small` is the OCS compact size |
| `style="padding: 6px;"` | `ocs__input medium` | Padding comes from the base class; `medium` is the standard size |
| Placeholder only | `<label for>` + matching `id` | Placeholders disappear when typing; labels stay visible (accessibility) |
