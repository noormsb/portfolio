## Popcorn

### Popcorn Hack 1: After-School Homework Plan

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan
is_weekday = True
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
````

This prints "Finish homework". Outer branch: `is_weekday` is True, so we're in the weekday path. Inner branch: `homework_finished` is False, so we land on the "not finished" message inside that path.

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan (homework_finished = True)
is_weekday = True
homework_finished = True

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
````

````python
# CODE_RUNNER: Popcorn Hack 1 - Test the after-school homework plan (is_weekday = False)
is_weekday = False
homework_finished = False

if is_weekday:
    if homework_finished:
        print("Free time")
    else:
        print("Finish homework")
else:
    print("Weekend plan")
````

All three outcomes reached: `is_weekday=True, homework_finished=False` -> "Finish homework"; `is_weekday=True, homework_finished=True` -> "Free time"; `is_weekday=False` -> "Weekend plan" (the inner check is never even reached on a weekend).

### Popcorn Hack 2: Repair the Student Account Check

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check (buggy version, run first)
logged_in = True
has_permission = False

if logged_in:
    print("Account found")
if has_permission:
    print("Page opened")
else:
    print("Permission needed")
````

This isn't actually nested: the `has_permission` check is a second, separate top-level `if`/`else` that runs on its own, not indented inside the `logged_in` block. That means it would run and print "Permission needed" even if `logged_in` were False — the permission check never actually depends on being logged in.

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check (fixed: nested + outer else)
logged_in = True
has_permission = False

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
````

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check (fixed, has_permission = True)
logged_in = True
has_permission = True

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
````

````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check (fixed, logged_in = False)
logged_in = False
has_permission = False

if logged_in:
    print("Account found")
    if has_permission:
        print("Page opened")
    else:
        print("Permission needed")
else:
    print("Log in first")
````

All three messages tested: logged in without permission -> "Account found" + "Permission needed"; logged in with permission -> "Account found" + "Page opened"; not logged in -> "Log in first" (and the permission check is skipped entirely, since it's now nested inside the `logged_in` branch).

### Popcorn Hack 3: Lunch and Sports Practice

````python
# CODE_RUNNER: Popcorn Hack 3 - Test the lunch and sports-practice plan
def lunch_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            print("Lunch packed - bring an extra snack for practice")
        else:
            print("Lunch packed - no extra snack needed")
    else:
        if practice_after_school:
            print("Buy lunch - also grab a snack for practice")
        else:
            print("Buy lunch - no extra snack needed")

# Test all four paths
lunch_plan(True, True)    # Prediction: Lunch packed - bring an extra snack for practice
lunch_plan(True, False)   # Prediction: Lunch packed - no extra snack needed
lunch_plan(False, True)   # Prediction: Buy lunch - also grab a snack for practice
lunch_plan(False, False)  # Prediction: Buy lunch - no extra snack needed
````

The same structure in JavaScript:
```javascript
if (lunchPacked) {
  if (practiceAfterSchool) { console.log("Lunch packed - bring an extra snack for practice"); }
  else { console.log("Lunch packed - no extra snack needed"); }
} else {
  if (practiceAfterSchool) { console.log("Buy lunch - also grab a snack for practice"); }
  else { console.log("Buy lunch - no extra snack needed"); }
}
```
And in College Board pseudocode:
```
IF (lunchPacked)
{
  IF (practiceAfterSchool) { DISPLAY("Lunch packed - bring an extra snack for practice") }
  ELSE { DISPLAY("Lunch packed - no extra snack needed") }
}
ELSE
{
  IF (practiceAfterSchool) { DISPLAY("Buy lunch - also grab a snack for practice") }
  ELSE { DISPLAY("Buy lunch - no extra snack needed") }
}
```
`lunch_packed` is checked first in all three forms — it's the outer condition, so `practice_after_school` is only ever checked after the lunch path is already decided.

## Homework

````python
# CODE_RUNNER: Homework Hack - Build a high school daily routine dashboard
def plan_day(student):
    if student["logged_in"]:
        if student["homework_finished"]:
            if student["lunch_packed"]:
                if student["practice_today"]:
                    return "Homework done, lunch packed - bring an extra snack for practice."
                else:
                    return "Homework done, lunch packed - head home after school, no practice today."
            else:
                if student["practice_today"]:
                    return "Homework done, buy lunch - grab a snack before practice too."
                else:
                    return "Homework done, buy lunch - head home after school, no practice today."
        else:
            return "Finish your homework before planning the rest of the day."
    else:
        return "Log in first to see your daily plan."

# Starter data, unmodified
student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}
print(plan_day(student))  # Prediction: Finish your homework before planning the rest of the day.
````

````python
# CODE_RUNNER: Homework Hack - test every other meaningful outcome
print(plan_day({"logged_in": True, "homework_finished": True, "lunch_packed": True, "practice_today": True}))
print(plan_day({"logged_in": True, "homework_finished": True, "lunch_packed": True, "practice_today": False}))
print(plan_day({"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": True}))
print(plan_day({"logged_in": True, "homework_finished": True, "lunch_packed": False, "practice_today": False}))
print(plan_day({"logged_in": False, "homework_finished": False, "lunch_packed": False, "practice_today": False}))
````

This dashboard nests three decisions deep: whether the student is logged in, then whether homework is finished, then the lunch/practice combination. The inner checks have to depend on the outer ones because they only make sense once the outer condition is already true — there's no point planning lunch and practice for a student who hasn't finished their homework yet, and no point checking homework at all for someone who isn't logged in. Each nested `if` only runs once its parent branch has already said "yes, keep going."
