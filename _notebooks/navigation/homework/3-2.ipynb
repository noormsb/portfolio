### Popcorn Checkpoint: Extend the level sequence

```python
# CODE_RUNNER: UESL 3.2 Popcorn - Extend the level sequence
level_stars = [6, 9, 12]

level_stars[1] = level_stars[1] + 3  # Python index 1 is AP index 2, the second level
level_stars.append(5)

print(level_stars)       # Prediction: [6, 12, 12, 5]
print(len(level_stars))  # Prediction: 4
```

```python
# CODE_RUNNER: UESL 3.2 Popcorn - Extend the level sequence (append 8 instead of 5)
level_stars = [6, 9, 12]

level_stars[1] = level_stars[1] + 3
level_stars.append(8)

print(level_stars)       # Prediction: [6, 12, 12, 8]
print(len(level_stars))  # Prediction: 4 (appending only ever adds one level)
```

AP index 2 = Python index 1: both select the second level. Updating `level_stars[1]` keeps the list length the same because it replaces an existing element, while `append()` increases the length by one because it adds a brand-new element at the end.

## MCQ

MCQ 3.2: 4/4 | answers: B,A,C,B

**Explanations:**
1. **1 - B (1):** Python's first index is 0, but AP CSP pseudocode lists start at index 1, so AP index 1 selects the first level.
2. **2 - A (0):** Replacing an existing element's value changes what is stored at that position but does not add or remove an element, so the list's length stays the same.
3. **3 - C (Add one element with value 5):** `APPEND(level_stars, 5)` adds a single new element with the value 5 to the end of the list; it does not change level five or add five levels.
4. **4 - B (It avoids a separate variable and sum edit for every level):** A list plus a traversal loop lets the same code work for any number of levels, so 100 levels do not need 100 separate variables and 100 edits to the total calculation.

## Homework

Each element in `level_stars` represents one level's collected stars, kept in level order — the first element is the first level's stars, the second is the second level's, and so on.

```python
# CODE_RUNNER: UESL 3.2 Homework - Grow the game's level collection
level_stars = [12, 18, 10]

# Guard: only update the second level if it actually exists (list has at least 2 elements)
if len(level_stars) >= 2:
    level_stars[1] = level_stars[1] + 5
level_stars.append(7)

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)       # Prediction: [12, 23, 10, 7]
print(len(level_stars))  # Prediction: 4
print(total_stars)       # Prediction: 52
```

```python
# CODE_RUNNER: UESL 3.2 Homework - Same guarded code on an empty collection
level_stars = []

# The guard skips the update since there's no second level yet, so this doesn't raise an IndexError
if len(level_stars) >= 2:
    level_stars[1] = level_stars[1] + 5
level_stars.append(7)

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(level_stars)       # Prediction: [7]
print(len(level_stars))  # Prediction: 1
print(total_stars)       # Prediction: 7
```

```python
# CODE_RUNNER: UESL 3.2 Homework - Total-only traversal on an empty collection
level_stars = []

total_stars = 0
for stars in level_stars:
    total_stars = total_stars + stars

print(len(level_stars))  # Prediction: 0
print(total_stars)       # Prediction: 0
```

A list plus this traversal avoids writing a new variable and a new total-calculation edit every time a level is added — the same `for` loop works whether there are 3 levels or 100. I'd suggest giving each level a clear label too (for example, a list of level names alongside the star counts, or a dictionary of name to star count), so a player or future maker doesn't have to remember what index 1 means.

## Design Thinking

**Need:** A game maker wants to add or reorder levels without rewriting a variable and a calculation every time.

**Goal:** Represent all of a game's level star counts with one list instead of one variable per level, and still be able to read, update, and total them correctly.

**Representation I considered:** I considered keeping the separate `level_one_stars`, `level_two_stars`, ... variables from 3.1, but that doesn't scale — adding a level means adding a new variable and hand-editing the total formula every time. A list with a loop avoids that.

**Prototype:** I built the level list, an update-by-index step, an append step, and a traversal that totals any number of levels.

**Test/revision:** I first wrote the second-level update without checking the list's length, and realized it would crash with an `IndexError` on an empty list. I revised it to check `len(level_stars) >= 2` before updating, then reran it on `[]` to confirm it no longer crashes.
