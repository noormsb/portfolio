## Popcorn

```python
# CODE_RUNNER: UESL 3.1 Popcorn - Keep the checkpoint
stars = 8
checkpoint_stars = stars
stars = stars + 4

print(stars)             # Prediction: 12
print(checkpoint_stars)  # Prediction: 8
```

The checkpoint holds the earlier value: `checkpoint_stars` copied `stars` while it was still 8, so adding 4 to `stars` afterward doesn't change what `checkpoint_stars` already stored.

## MCQ

MCQ 3.1: 4/4 | answers: B,A,C,B

**Explanations:**
1. **1 - B (string):** `"007"` is wrapped in quotes and contains leading zeros, so it must be stored as text/string rather than an integer.
2. **2 - A (8):** `checkpoint` was assigned the value of `stars` when `stars` was 8. Updating `stars` to 12 afterward does not retroactively change `checkpoint`.
3. **3 - C (←):** In College Board AP CSP pseudocode, `←` represents assignment, while `=` tests equality.
4. **4 - B (When reassigned/recalculated):** Stored totals do not automatically track variable updates; they only update when you explicitly recalculate the formula and reassign the variable.

## Homework

```python
# CODE_RUNNER: UESL 3.1 Homework - Build the star counter
game_title = "UESL Star Trail"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10
total_stars = level_one_stars + level_two_stars + level_three_stars

print(game_title)      # Prediction: UESL Star Trail
print(reduced_motion)  # Prediction: True
print(total_stars)     # Prediction: 40

# Add a bonus of 5 to level two
level_two_stars = level_two_stars + 5
print(total_stars)     # Prediction: 40 (stored total hasn't been recalculated yet)

total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)     # Prediction: 45 (recalculated after the bonus)
```

```python
# CODE_RUNNER: UESL 3.1 Homework - Restart with no bonus
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10
total_stars = level_one_stars + level_two_stars + level_three_stars

print(total_stars)  # Prediction: 40

# Bonus of 0 this time
level_two_stars = level_two_stars + 0
print(total_stars)  # Prediction: 40

total_stars = level_one_stars + level_two_stars + level_three_stars
print(total_stars)  # Prediction: 40 (all three totals stay 40 with no bonus)
```

Reduced motion should be a player's choice, not a fixed setting, because motion sensitivity varies from player to player — some players need less on-screen motion to play comfortably (for reasons like motion sickness or attention), while others don't mind it. Letting each player toggle it respects that difference instead of forcing one setting on everyone.

## Tests

Both tests confirm that a stored total only changes when it is explicitly recalculated and reassigned — adding a bonus to a level variable does not, by itself, update a total that was already calculated and stored.

## Design Thinking

**Need:** A player wants to trust that a displayed total actually reflects what they've earned, instead of the game silently showing a stale number after something changes.

**Goal:** Model a simple game state (a name, a reduced-motion setting, and level star totals) so I can trace exactly when a value changes versus when a stored total needs to be recalculated.

**Representation I considered:** I considered reusing one variable for both the "live" count and the "checkpoint" count, but that would overwrite the earlier value entirely, so I kept them as two separate variables (`stars` and `checkpoint_stars`) instead.

**Prototype:** I built the star-counter checkpoint example, then reused the same initialize/update/recalculate pattern for the level totals in the homework.

**Test/revision:** I first assumed printing `total_stars` right after adding the bonus would already show 45, but running it showed 40 — I'd forgotten `total_stars` is a stored value, not a live formula, so I added an explicit recalculation line before printing it again.
