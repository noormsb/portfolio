## Popcorn

### Popcorn Hack 1: The Silent Failure

```python
# CODE_RUNNER: 3.6 Popcorn 1 - Run it first. Then make both outcomes print.
amount_requested = 0
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
```

Running this unchanged prints nothing. That's not an error — it's a bug: the `if` has no `else`, so when `amount_requested > 100` is False, the program silently does nothing instead of saying the message looks fine.

```python
# CODE_RUNNER: 3.6 Popcorn 1 - Run it first. Then make both outcomes print. (fixed)
amount_requested = 0

# 1 condition -> 2 outcomes -> 2 answers
if amount_requested > 100:
    print("This message is asking for a large amount of money.")
else:
    print("This message is not asking for a large amount of money.")
```

### Popcorn Hack 2: The Chain in the Wrong Order

```python
# CODE_RUNNER: 3.6 Popcorn 2 - Run it first. Then fix the order.
risk_score = 9
if risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
else:
    print("Nothing suspicious in this message.")
```

For `risk_score = 9` — the worst possible score — this wrongly tells the person "Be a little careful, but this is probably fine," because the first branch (`>= 3`) catches every score of 3 or higher before the chain ever reaches the later checks. The `elif risk_score >= 6` and `elif risk_score >= 9` branches can never run **for any score at all**: any score that would satisfy them already satisfied `>= 3` first, so `elif` never even asks.

```python
# CODE_RUNNER: 3.6 Popcorn 2 - Run it first. Then fix the order. (fixed: highest threshold first)
risk_score = 9

# 3 thresholds -> 4 outcomes -> 4 answers
if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")
```

```python
# CODE_RUNNER: 3.6 Homework - test the extremes: a completely clean message
message = "Hi, are we still on for Sunday?"
amount_requested = 20
known_contact = True
risk_score = 0

if "gift card" in message:
    risk_score += 4
if amount_requested > 100:
    risk_score += 3
if not known_contact:
    risk_score += 2

if risk_score >= 9:
    print("Do not reply. This is almost certainly a scam.")
elif risk_score >= 6:
    print("This looks risky. Check with someone you trust before replying.")
elif risk_score >= 3:
    print("Be a little careful, but this is probably fine.")
else:
    print("Nothing suspicious in this message.")

print("Risk score:", risk_score)  # Prediction: 0 (no signal is tripped)
```

```python
# CODE_RUNNER: 3.6 Homework - reachability check for the middle two branches
def score_and_report(message, amount_requested, known_contact):
    risk_score = 0
    if "gift card" in message:
        risk_score += 4
    if amount_requested > 100:
        risk_score += 3
    if not known_contact:
        risk_score += 2

    if risk_score >= 9:
        advice = "Do not reply. This is almost certainly a scam."
    elif risk_score >= 6:
        advice = "This looks risky. Check with someone you trust before replying."
    elif risk_score >= 3:
        advice = "Be a little careful, but this is probably fine."
    else:
        advice = "Nothing suspicious in this message."
    return risk_score, advice

# gift card mentioned + unknown sender, but a small amount -> should land in the ">= 6" branch
print(score_and_report("send a gift card please", 20, False))   # Prediction: (6, 'This looks risky...')

# only the amount signal trips -> should land in the ">= 3" branch
print(score_and_report("Hi, are we still on for Sunday?", 200, True))  # Prediction: (3, 'Be a little careful...')
```

**Reachability proof — one input set per branch:**

- `risk_score >= 9`: gift card mention + amount 500 (>100) + unknown sender -> score 9 -> "Do not reply."
- `risk_score >= 6`: gift card mention + unknown sender, but amount only 20 (not >100) -> score 6 -> "This looks risky."
- `risk_score >= 3`: amount 200 (>100) only, known contact, no gift card mention -> score 3 -> "Be a little careful."
- `else`: clean message, amount 20, known contact -> score 0 -> "Nothing suspicious."

Every branch has at least one real input that reaches it, so none of the four answers is dead code.
