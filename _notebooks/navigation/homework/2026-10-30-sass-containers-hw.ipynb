## Homework Hack: Parade Hub Page
One `ocs__container` holding one `ocs__card`: title, description, a `cols-3` standard grid with header, accent, and muted cells, a wrapped table (header row + two body rows), and a callout. No inline styles, no made-up classes.

````python
%%html
<!-- UI_RUNNER: Containers Homework -->

<div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade</h2>
    <p class="ocs__description">Everything you need for the Poway Scripps Rotary Parade in one place.</p>

    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell ocs__grid-cell--header">Parade Details</div>
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00 AM</div>
      <div class="ocs__grid-cell ocs__grid-cell--muted">Parade floats</div>
    </div>

    <div class="ocs__table-wrap">
      <table class="ocs__table">
        <thead>
          <tr><th>Entry</th><th>Status</th></tr>
        </thead>
        <tbody>
          <tr><td>Float registration</td><td>Open</td></tr>
          <tr><td>Volunteer sign-up</td><td>Open</td></tr>
        </tbody>
      </table>
    </div>

    <p class="ocs__callout">Arrive by 8:30 AM to find a good spot along Poway Road.</p>
  </div>
</div>
````

## How it fits

```text
ocs__container            ← one per page, sets the readable width
└── ocs__card             ← framed group
    ├── ocs__section-title + ocs__description
    ├── ocs__grid --standard cols-3
    │     ├── --header   Parade Details
    │     ├── cell       Location
    │     ├── --accent   Start time
    │     └── --muted    Parade floats
    ├── ocs__table-wrap → ocs__table (thead + 2 body rows)
    └── ocs__callout
```

Tested at narrow and wide widths: the grid collapses to one column on phones because the OCS grid classes handle it.
