## How it fits

```text
ocs__container            ← one per page, sets the readable width
└── ocs__card             ← framed group
    ├── ocs__section-title + ocs__description
    ├── ocs__grid --standard cols-3
    │     ├── --header   Parade Details
    │     ├── cell       Location
    │     ├── --accent   Start time
    │     └── --muted    Parade floats
    ├── ocs__table-wrap → ocs__table (thead + 2 body rows)
    └── ocs__callout
```

Tested at narrow and wide widths: the grid collapses to one column on phones because the OCS grid classes handle it.

## The SASS behind my containers

The page uses only OCS container grammar. This is the SASS behind those classes, based on OCS `elements/containers/common.scss`: `ocs__container` sets a readable max width and centers the page, a **`panel` mixin** gives the card and the callout the same look without repeating code, `ocs__table-wrap` lets the table scroll on small screens, and a **boolean + `@if` / `@else`** controls the callout's accent border.

```scss
// ---- Settings ----
$container-max-width: 900px;
$container-padding: 2rem 1.5rem;
$card-radius: 0.75rem;
$callout-has-border: true;   // boolean: left accent bar on callouts

// ---- Mixin: shared "panel" look reused by card and callout ----
@mixin panel($radius: $card-radius) {
  background: var(--panel);
  border: 1px solid var(--ui-border);
  border-radius: $radius;
  padding: 1.25rem;
}

// One page container: sets a readable width and centers the page
.ocs__container {
  max-width: $container-max-width;
  margin: 0 auto;
  padding: $container-padding;
  color: var(--pref-text-color);

  &, * { box-sizing: border-box; }
}

// Card: the box that holds the Parade hub content
.ocs__card {
  @include panel;
}

.ocs__section-title { font-size: 1.5rem; font-weight: 600; margin: 0 0 0.25rem; }
.ocs__description   { opacity: 0.8; margin: 0 0 1rem; }

// Table: the wrap scrolls sideways on small screens instead of breaking the layout
.ocs__table-wrap { overflow-x: auto; }
.ocs__table {
  width: 100%;
  border-collapse: collapse;

  th, td { padding: 0.5rem 0.75rem; border-bottom: 1px solid var(--ui-border); text-align: left; }
  th     { font-weight: 600; }
}

// Callout: the one-line note at the end
.ocs__callout {
  @include panel(0.5rem);

  @if $callout-has-border {
    border-left: 4px solid var(--pref-accent-color);
  } @else {
    border-left: none;
  }
}
```
