### Popcorn Checkpoint: Expand the catalog

```python
# CODE_RUNNER: Library 3.17 Popcorn - Expand the catalog
import math

n = 8
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print(n)              # Prediction: 8
print(linear_checks)  # Prediction: 8
print(binary_checks)  # Prediction: 4
```

```python
# CODE_RUNNER: Library 3.17 Popcorn - Expand the catalog (changed to n = 16)
import math

n = 16
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print(n)              # Prediction: 16
print(linear_checks)  # Prediction: 16
print(binary_checks)  # Prediction: 5
```

Recommendation: doubling the catalog from 8 to 16 books doubles the linear-search worst case (8 -> 16), but binary search only grows by one more check (4 -> 5), since it halves the remaining catalog each time instead of scanning one entry at a time.

## MCQ

MCQ 3.17: 4/4 | answers: A,B,B,C

**Explanations:**
1. **1 - A (a sorted catalog with direct access to entries):** Binary search only works by repeatedly halving a sorted range, which requires both sorted order and direct access to the middle entry.
2. **2 - B (work and memory as input grows):** Comparing two correct solutions means comparing their efficiency, the time and memory they use as n grows, not code length or variable names.
3. **3 - B (400):** A quadratic model's work scales with n^2, so doubling n from 10 to 20 multiplies the work by 2^2 = 4, giving 100 x 4 = 400.
4. **4 - C (can be useful without guaranteeing the best list):** A heuristic like reading short books first is a useful shortcut, but it is not guaranteed to produce the optimal result.

```python
# Adapt Example C for a catalog of 12 books: display the book count, visits, and ordered pairs
n = 12
visits = n
pairs = n * n

print("Books:", n)      # Prediction: 12
print("Visits:", visits)  # Prediction: 12
print("Pairs:", pairs)    # Prediction: 144
```

```python
# Double the catalog to 24 books and compare growth
n = 24
visits = n
pairs = n * n

print("Books:", n)         # Prediction: 24
print("Visits:", visits)  # Prediction: 24
print("Pairs:", pairs)    # Prediction: 576
# Doubling n from 12 to 24 doubles visits (12 -> 24) but quadruples pairs (144 -> 576),
# since visits grow linearly with n while pairs grow with n * n.
```

```python
# Use checkpoint B's formula for a catalog of 1,000 books to recommend a search method
import math

n = 1000
linear_checks = n
binary_checks = 0 if n == 0 else math.floor(math.log2(n)) + 1

print("Linear checks:", linear_checks)  # Prediction: 1000
print("Binary checks:", binary_checks)  # Prediction: 10
```

Recommendation: for repeated searches of an already-sorted catalog of 1,000 books, binary search is far better (10 worst-case checks versus 1,000 for linear search). If the catalog is unsorted, it must be sorted first, which costs extra time, and keeping a separate sorted copy costs extra memory, so the recommendation only holds once sorting is already paid for or the catalog stays sorted.

## Design Thinking

**Need:** A librarian wants to find a book quickly even as the catalog grows into the thousands, without guessing at how much work each approach actually takes.

**Goal:** Compare linear and binary search by counting checks instead of judging by code length, so I can recommend a method based on real growth, not a guess.

**Methods I considered:** Linear search (simple, no sorting required, but worst case grows with n) and binary search (needs a sorted catalog with direct index access, but grows much more slowly).

**Prototype:** I built small calculators that count linear and binary checks for a given catalog size, plus a separate visits/pairs example to see quadratic growth.

**Test/revision:** I first ran the checkpoint with n = 8 and assumed both counts would grow the same way as the catalog got bigger; after rerunning with n = 16 I saw linear checks doubled (8 -> 16) while binary checks only went up by one (4 -> 5), so I revised my explanation to say they grow at very different rates instead of "about the same."
